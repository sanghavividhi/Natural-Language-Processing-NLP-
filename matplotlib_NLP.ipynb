```python
import matplotlib.pyplot as plt
month=['Jan','Feb','Mar','Apr']
user=[1200,1500,1450,1800]
plt.plot(month,user,marker='o',color='blue')
plt.title('Active users by month')
plt.show()
```

```python
ad_spend=[100,200,300,400,500]
conversions=[15,25,40,35,60]
plt.scatter(ad_spend,conversions,color='blue')
plt.xlabel('Ad spend')
plt.ylabel('Conversions')
plt.title('Ad spend vs Conversions')
plt.show()
```

```python
data={
    'House_Age':[5,12,7,20,15,8,25,10,18,6],
    'Bedrooms':[2,3,2,4,3,2,5,3,4,2],
    'Square_Footage':[1200,1600,1300,2400,1800,1400,2800,1700,2200,1250],
    'Price':[180000,240000,200000,350000,280000,210000,420000,260000,330000,190000]

}
df=pd.DataFrame(data)
correlation_matrix=df.corr(numeric_only=True)
plt.figure(figsize=(7,5))
sns.heatmap(correlation_matrix,annot=True,cmap='magma',fmt='.2f')
plt.title('Matrix Correlation for Reature Selection')
plt.show()
```

```python
#task2

import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("region_sales.csv")
df
```

```python
region_sales=df.groupby('Region')['Units_Sold'].sum()
plt.figure(figsize=(8,5))
plt.bar(region_sales.index,region_sales.values,color='green')
plt.title('Total Units Sold by Region')
plt.xlabel('Region')
plt.ylabel('Total Units Sold')
plt.show()
```
