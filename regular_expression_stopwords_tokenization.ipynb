```python
#regular_expression,stopwords,tokenization

import re

text="Check this out !! visit https://nlp.course.com NOW !!! call 7874503625 #offer"
text= re.sub(r'https?://\S+', '', text)
print(text)
```

```python
text="Check this out !! visit https://nlp.course.com NOW !!! call 7874503625 #offer"

text= re.sub(r'https?://\S+', '', text)
text= re.sub(r'https?://+','', text) #remove urls
text=re.sub(r'\d+','',text) #remove numbers
text=re.sub(r'[^a-zA-Z\s]','',text) #remove punctuation/symbols
text=re.sub(r'/s+','',text) #collapse spaces
print(text)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

tokens=word_tokenize(text)

filtered=[word for word in tokens if word not in stop_words]
print(filtered)
```

```python
def clean_text(text):
  text= re.sub(r'https?://+','', text)
  text=re.sub(r'[^a-zA-Z\s]','',text).lower()
  tokens=word_tokenize(text)
  return [word for word in tokens if word not in stop_words]

print(clean_text("Check this out !! visit https://nlp.course.com NOW !!! call 7874503625 #offer"))
```

```python
from nltk.tokenize import sent_tokenize,word_tokenize

text="Tokenization is fun! It helps NLP models understand text."
sentences=sent_tokenize(text)
word=word_tokenize(text)

print("Sentences:",sentences)
print("Words:",word)

length=len(sentences)
print("Number of sentences:",length)
length=len(word)
print("Number of words:",length)
```

```python
import nltk
from nltk.stem import PorterStemmer,SnowballStemmer,LancasterStemmer
from nltk.tokenize import word_tokenize

porter=PorterStemmer()
snowball=SnowballStemmer('english')
lencaster=LancasterStemmer()
words=['studies','studying','running','cries','easily','fairly']

for word in words:
  print(word,'->',porter.stem(word),snowball.stem(word),lencaster.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer=WordNetLemmatizer()

#pos='v' treats the word as a verb
#pos='a' would treat it as an objective
words=['studies','studying','running','cries','easily','fairly']
for word in words:
  print(word,'->',lemmatizer.lemmatize(word,pos='v'))
```

```python
import spacy

nlp=spacy.load('en_core_web_sm')

doc=nlp("The striped bats are hanging on their fest for best and were crying")

for token in doc:
  print(token.text,'->',token.lemma_)
```
