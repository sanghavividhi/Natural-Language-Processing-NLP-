```python
import pandas as pd
```

```python
df=pd.read_csv('nba.csv')
print(df)
```

```python
team=df.groupby('Team')
print(team.first())
```

```python
grouping=df.groupby(['Team','Position'])
print(grouping.first())
```

```python
team1=df.groupby('Team')['Salary'].max()
print(team1)
```

```python
team2=df.groupby('Team')['Salary'].sum()
print(team2)
```

```python
aggregated_data=df.groupby(['Team','Position']).agg(
    total_salary=('Salary','sum'),
    avg_salary=('Salary','mean'),
    player_count=('Name','count')
)
print(aggregated_data)
```

```python
#pandas_task1
data=pd.read_csv('customer_reviews.csv')
data
```

```python
print(data.info())
```

```python
data=data.drop_duplicates()
print(data)
```

```python
data['Age']=data['Age'].fillna(data['Age'].mean())
```

```python
data.isnull().sum()
```

```python
rating=data[data['Rating'] == 5]
print(rating)
```

```python
data['Review_Length']=data['Review_Text'].str.len()

avg_review_length = data.groupby('Product_Category')['Review_Length'].mean()
print(avg_review_length)
```

```python
#task2
data1=pd.read_csv("customer_complex_data.csv")
print(data1)
```

```python
print(data1.duplicated)
```

```python
new_data=data1.drop_duplicates()
print(new_data)
```

```python
print(data1)
```

```python
average_age=data1['Age'].mean()
data1['Age'].fillna(average_age,inplace=True)
print(data1)
```

```python
data1.loc[data1['Name'] == "Kabir","Spending_Score"] = 0
print(data1)
```

```python
data1['Join_Date']=pd.to_datetime(data1['Join_Date'],errors='coerce')
print(data1)
```

```python
data1=data1.dropna(subset=['Join_Date'])
print(data1)
```

```python
data1.loc[data1['Age'] < 0,'Age']=25
print(data1)
```

```python
result=data1.groupby('Category').agg(
    Customer_Count=('Customer_ID','count'),
    Mean_Spending_Score=('Spending_Score','mean'),
    Total_Spending_Score=('Spending_Score','sum')
)
print(result)
```
